```python
feature = iris_data.iloc[:,0:4]
label = iris_data.iloc[:,4]

print("Features : ")
print(feature)

print("Label : ")
print(label)
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(feature,label, test_size=0.3,random_state=0)


print("Number of trained data", x_train.shape[0])
print("Number of data test" , y_test.shape[0])
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors = 5,weights = "uniform")
## N_neighbours to select the radius / The Variable is the model

knn.fit(x_train,y_train)
## Feed the model

train_acc = knn.score(x_train,y_train)

print("The Accuracy of KNN Classifier on Training data is : {:.3f}".format(train_acc))
```

```python
knn_test_acc = knn.score(x_test,y_test)
print("The Accuracy of KNN Classifier on Testing data is : {:.3f}".format(knn_test_acc))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

disp = ConfusionMatrixDisplay.from_estimator(knn,x_test,y_test)

disp.ax_.set_title("Confusion Matrix")
```

```python
from subprocess import check_output
from sklearn import tree
from sklearn.pipeline import make_pipeline
from sklearn import preprocessing
from sklearn.model_selection import ShuffleSplit
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
```

```python
data = pd.read_csv("iris.data")
feature = data.iloc[:,0:4]
label = data.iloc[:,4]
```

```python
from sklearn.neighbors import KNeighborsClassifier

n_range = list(range(1,10))
weight_list = ["uniform","distance"]
param_grid = {"n_neighbors": n_range,"weights":weight_list}

grid= GridSearchCV(estimator= KNeighborsClassifier(),param_grid = param_grid,cv=3,scoring = "accuracy",refit = True)
knn_model = make_pipeline(preprocessing.StandardScaler(), grid)
knn_model.fit(feature,label)

print("accuracy of the tuned model: %.4f " %grid.best_score_)
print(grid.best_params_)
```

```python
import pickle

with open("knn_model.pickle","wb") as f:
    pickle.dump(knn_model,f)
```

```python
# Loading the Model
with open("knn_model.pickle","rb") as f:
    model = pickle.load(f)

# Predict new Data
new_data=[5.0,2.5,1.0,1.2]
x = np.array([new_data])

prediction= model.predict(x)
print("Prediciton of Species: {}". format(prediction))
```
